```python
# Bosluqlari NaN ile evezleyirik ve sutunu eded tipine (float) ceviririk
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
```

```python
df.info()
```

```python
# totalcharges xanasinda null olduguna gore onlari monthlycharges deyerleri ile evez edirik ki datada null olmasin
df['TotalCharges'] = df['TotalCharges'].fillna(df['MonthlyCharges'])
```

```python
df.isnull().sum()
```

```python
#MultipleLines sutununda unikal deyerli gormek ucun
df['MultipleLines'].unique()
```

```python
df['MultipleLines'] = df['MultipleLines'].replace('No phone service', 'No')
```

```python
df['MultipleLines'].unique()
```

```python
sutunlar = [
    'OnlineSecurity', 'OnlineBackup', 
    'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies',
]

#bu sutunlar 2 den cox deyeri olan sutunlardi onlari cap edirik sonra evez etmeye gore replace
for sutun in sutunlar:
    print(f"{sutun}: {df[sutun].unique()}")
```

```python
sutunlar = [
    'MultipleLines', 'OnlineSecurity', 'OnlineBackup', 
    'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies'
]

#tek tek etmek evezine dovr qurub bir dovrde butun sutunlarda gezib 'No internet service'-i ve No Phone services-i No ile evez edir
for sutun in sutunlar:
    df[sutun] = df[sutun].replace({'No internet service': 'No', 'No phone service': 'No'})
```

```python
sutunlar = [
    'MultipleLines', 'OnlineSecurity', 'OnlineBackup', 
    'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies'
]

#daha sonra yeniden sutunda 2 cur melumat oldugunu yoxluyuruq
for sutun in sutunlar:
    print(f"{sutun}: {df[sutun].unique()}")
```

```python
# Əvvəlcə internet və telefon xidməti olmayanları "No" edirik
sutunlar = [
    'Partner', 'Dependents', 'PhoneService', 'MultipleLines', 'InternetService', 
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 
    'StreamingTV', 'StreamingMovies', 'PaperlessBilling', 'Churn'
]

for sutun in sutunlar:
    if sutun in df.columns:
        # Həm bərabərləşdirməni, həm də 1-0 keçidini tək sətirdə təhlükəsiz edirik:
        df[sutun] = df[sutun].replace({'No internet service': 'No', 'No phone service': 'No'})
        df[sutun] = df[sutun].map({'Yes': 1, 'No': 0}).fillna(0).astype(int)
```

### Tezis datadan cixarilan esas faktlar

correlyasiyadan cixan esas asliliq bunlardi hansi sutunun hansi sutunla bagli olub olmamasini bele tapmaq olar

1) 0.83-tenure ve totalCharges (0.83): Datadaki en guclu elaqedir. Musterinin sirketde qaldigi ay sayi (tenure) artdiqca, sirkete odeyiyi umumi mebleg (TotalCharges) demek olar ki, duz mutenasib sekilde yukselir.

2) 0.65-MonthlyCharges ve TotalCharges (0.65): Aylig odenis meblegi yuksek olan musterilerin tebii olaraq umumi gelire verdiyi tohfe de daha boyuk olur.

3) Her ikisi 0.63-MonthlyCharges ve StreamingTV / StreamingMovies (Her ikisi 0.63): Musteriler paketlerine TV ve Kino elave etdikce, onlarin aylig odenis haqqi keskin sekilde artir.

4) 0.53-StreamingTV ve StreamingMovies (0.53): Bu iki xidmet bir-biri ile six baglidir. Yeni, platformada TV yayimlarini aktivlesdiren musteriler cox boyuk ehtimalla kino paketlerini de birlikde alirlar.


1) tenure ve Churn (-0.35): Datadaki en guclu ters mutenasib elaqedir. Musterinin sirketde qaldigi ay sayisi (tenure) artdiqca, sistemden cixma/itki ehtimali (Churn) azalir. Bu, yeni musterilerin sirketi terk etmeye en cox meylli grup oldugunu riyazi olaraq subut edir.

2) Dependents ve SeniorCitizen (-0.21): Yasli musterilerin (SeniorCitizen) himayesinde yetkinlik yasina catmayan ve ya baxima mohtac sexslerin (Dependents) olma ehtimali daha asagidir.

3) TotalCharges ve Churn (-0.20): Sirkete umumi olaraq cox pul qazandirmis musterilerin sistemden cixma riski daha azdir.
